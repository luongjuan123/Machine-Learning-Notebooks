```python
import numpy as np

def sigmoid(x):
    return 1/(1 + np.exp(-x))

class LinearModel(object):
    """Base class for linear models"""
    def __init__(self, step_size=0.2, max_iter=100, eps=1e-6, theta_0=None, verbose=True):
        self.step_size = step_size
        self.max_iter = max_iter
        self.eps = eps
        self.theta = theta_0
        self.verbose = verbose

    def fit(self, X, y):
        pass

    def predict(self, X):
        pass

class BatchGradientAscentLogisticRegression(LinearModel):

    def fit(self, X, y):
        m, n = X.shape
        self.theta = np.zeros(n)

        for _ in range(self.max_iter):
            theta_old = np.copy(self.theta)
            h = sigmoid(X @ self.theta)
            gradient = X.T @ (y - h) / m
            self.theta += self.step_size * gradient

            if np.linalg.norm(self.theta - theta_old, ord = 1) < self.eps:
                break

    def predict(self, X):
        return sigmoid(X @ self.theta)

if __name__ == "__main__" :
    
    m, n, t = map(int, input().split())

    X_train = np.array(input().split(), dtype=float).reshape(m, n)
    y_train = np.array(input().split(), dtype=float)
    X_pred = np.array(input().split(), dtype=float).reshape(t, n)

    model = BatchGradientAscentLogisticRegression()

    model.fit(X_train, y_train)

    y_pred = model.predict(X_pred)

    print(*[f"{x:.4f}" for x in model.theta])
    print(*[f"{x:.4f}" for x in y_pred])
```

```python
import random
import json

random.seed(42)

NUM_TESTS = 100

tests = []

for _ in range(NUM_TESTS):

    # n <= 20 để brute force
    n = random.randint(1, 20)

    a = [random.randint(1, 100) for _ in range(n)]

    # 70% có nghiệm
    if random.random() < 0.7:
        mask = random.randint(1, (1 << n) - 1)
        s = 0
        for i in range(n):
            if (mask >> i) & 1:
                s += a[i]
    else:
        s = random.randint(1, sum(a) + 100)

    ####################
    # Brute force answer
    ####################

    INF = 10**18
    ans = INF

    for mask in range(1, 1 << n):

        cur_sum = 0
        cnt = 0

        for i in range(n):
            if (mask >> i) & 1:
                cur_sum += a[i]
                cnt += 1

        if cur_sum == s:
            ans = min(ans, cnt)

    if ans == INF:
        ans = -1

    ####################
    # Build testcase
    ####################

    input_data = "\n".join([
        f"{n} {s}",
        " ".join(map(str, a))
    ])

    output_data = str(ans)

    tests.append({
        "input": input_data,
        "output": output_data
    })

####################
# Save
####################

with open("testcases.json", "w", encoding="utf-8") as f:
    json.dump(tests, f, indent=4, ensure_ascii=False)

print("Generated", NUM_TESTS, "test cases.")
```
